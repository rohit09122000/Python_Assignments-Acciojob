### Q3
Create an empty set and add numbers from 1 to 5 using a loop.

````python
# write your code here
empty_set = set()
for i in range(1,6):
    empty_set.add(i)
    print(empty_set)
````

### Q5
Given:
```python
A = {2, 4, 6, 8}
B = {4, 8, 12}
```
Find Union, Intersection, A-B

````python
# write your code here
A = {2, 4, 6, 8}
B = {4, 8, 12}
print(A.union(B))
print(A.intersection(B))
print(A.difference(B))
````
